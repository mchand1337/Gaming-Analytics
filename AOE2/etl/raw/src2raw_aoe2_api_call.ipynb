```python
import requests
import pandas as pd
from pyspark.sql import SparkSession

# 1. API call to get civilizations
url = 'https://age-of-empires-2-api.herokuapp.com/api/v1/civilizations'
headers = {'Accept': 'application/json'}

response = requests.get(url, headers=headers)
response.raise_for_status()  # throws an error if status is not 200

# 2. Normalize JSON into Pandas DataFrame
data = response.json()
civilizations = data.get("civilizations", [])  # extract the list of civs
pdf = pd.json_normalize(civilizations)

# 3. Convert to Spark DataFrame
spark = SparkSession.builder.getOrCreate()
sdf = spark.createDataFrame(pdf)

# # 4. Write to Databricks SQL Warehouse (Delta Table)
# sdf.write.format("delta") \
#     .mode("overwrite") \
#     .option("overwriteSchema", "true") \
#     .saveAsTable("aoe2_civilizations")
```
